```python
!pip install pandas beautifulsoup4 requests

import time
import requests
from bs4 import BeautifulSoup
import pandas as pd

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}

url = "http://quotes.toscrape.com/"

print("Scraping shuru ho rahi hai...")

try:
    response = requests.get(url, headers=headers, timeout=10)
    response.raise_for_status()
except requests.exceptions.RequestException as e:
    print(f"Error aa gaya: {e}")
    exit()

soup = BeautifulSoup(response.text, 'html.parser')
data = []

quotes_divs = soup.find_all('div', class_='quote')

for q in quotes_divs:
    text = q.find('span', class_='text').get_text()
    author = q.find('small', class_='author').get_text()

    data.append({
        'Quote': text,
        'Author': author
    })
    time.sleep(1)

df = pd.DataFrame(data)
df.to_csv('output.csv', index=False, encoding='utf-8')
print("Data successfully 'output.csv' mein save ho gaya hai!")

print("\n--- SUMMARY STATISTICS ---")
print(f"Total quotes scrape kiye gaye: {len(df)}")
print(f"Unique authors ki sankhya: {df['Author'].nunique()}")
```
